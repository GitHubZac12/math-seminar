# 05 — Spatial effects and adjacency

This notebook visualizes the spatial adjacency graph and residual spatial effects from the Bayesian Spatial Hierarchical Model.

```python

from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# This lets the notebook work whether it is launched from the project root or from notebooks/.
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
OUTPUTS = PROJECT_ROOT / "outputs"
FIGURES = OUTPUTS / "figures"
FIGURES.mkdir(parents=True, exist_ok=True)

pd.set_option("display.max_columns", 120)
pd.set_option("display.max_rows", 100)


def read_json(path):
    path = Path(path)
    if not path.exists():
        return {}
    with path.open("r", encoding="utf-8") as f:
        return json.load(f)


def require_file(path):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"Missing expected file: {path}")
    return path


def load_csv(relative_path, **kwargs):
    path = require_file(OUTPUTS / relative_path)
    return pd.read_csv(path, **kwargs)


def savefig(name):
    path = FIGURES / name
    plt.tight_layout()
    plt.savefig(path, dpi=160, bbox_inches="tight")
    print(f"Saved figure: {path}")

print(f"Project root: {PROJECT_ROOT}")
print(f"Outputs folder: {OUTPUTS}")

import geopandas as gpd
import plotly.express as px
```

## Load spatial outputs

```python

adj_meta = read_json(OUTPUTS / "spatial_adjacency_metadata.json")
neighbors = load_csv("spatial_neighbors.csv")
adj = load_csv("spatial_adjacency_matrix.csv", index_col=0)
effects = load_csv("bayes_spatial/bayesian_spatial_effect_summary.csv")
gdf = gpd.read_file(OUTPUTS / "risk_map_with_bayesian_spatial.geojson")

adj.index = adj.index.astype(str)
adj.columns = adj.columns.astype(str)
effects["geo_id"] = effects["geo_id"].astype(str)
gdf["geo_id"] = gdf["geo_id"].astype(str)

print(adj_meta)
print("Adjacency shape:", adj.shape)
print("Neighbors:", neighbors.shape)
effects.head()
```

## Neighbor-count distribution

```python

degree = adj.sum(axis=1).rename("n_neighbors").reset_index().rename(columns={"index": "geo_id"})

display(degree["n_neighbors"].describe())

fig, ax = plt.subplots(figsize=(7, 5))
ax.hist(degree["n_neighbors"], bins=range(int(degree["n_neighbors"].min()), int(degree["n_neighbors"].max()) + 2))
ax.set_xlabel("Number of neighboring municipalities")
ax.set_ylabel("Number of municipalities")
ax.set_title("Adjacency degree distribution")
savefig("05_adjacency_degree_distribution.png")
plt.show()
```

## Spatial effects table

```python

effects["combined_area_effect_width"] = effects["combined_area_effect_ci_upper"] - effects["combined_area_effect_ci_lower"]
effects.sort_values("combined_area_effect_mean", ascending=False).head(15)
```

## Combined area effect interval plot

```python

plot_df = effects.reindex(effects["combined_area_effect_mean"].abs().sort_values(ascending=False).index).head(20).copy()
plot_df = plot_df.sort_values("combined_area_effect_mean")

x = plot_df["combined_area_effect_mean"]
xerr = np.vstack([x - plot_df["combined_area_effect_ci_lower"], plot_df["combined_area_effect_ci_upper"] - x])

fig, ax = plt.subplots(figsize=(9, 7))
ax.errorbar(x, plot_df["geo_id"], xerr=xerr, fmt="o", capsize=3)
ax.axvline(0, linestyle="--")
ax.set_xlabel("Combined spatial + unstructured area effect")
ax.set_ylabel("Municipality geo_id")
ax.set_title("Largest residual area effects")
savefig("05_combined_area_effect_intervals.png")
plt.show()
```

## Map combined residual area effect

```python

map_df = gdf.merge(effects, on="geo_id", how="left", suffixes=("", "_effect"))
fig = px.choropleth(
    map_df,
    geojson=map_df.__geo_interface__,
    locations="geo_id",
    featureidkey="properties.geo_id",
    color="combined_area_effect_mean",
    hover_name="geo_id",
    hover_data=["combined_area_effect_mean", "combined_area_effect_ci_lower", "combined_area_effect_ci_upper"],
    title="Combined residual area effect from Bayesian spatial model",
)
fig.update_geos(fitbounds="locations", visible=False)
fig.update_layout(margin={"l": 0, "r": 0, "t": 50, "b": 0})
fig
```

## Map residual-effect uncertainty

```python

map_df["combined_area_effect_width"] = map_df["combined_area_effect_ci_upper"] - map_df["combined_area_effect_ci_lower"]
fig = px.choropleth(
    map_df,
    geojson=map_df.__geo_interface__,
    locations="geo_id",
    featureidkey="properties.geo_id",
    color="combined_area_effect_width",
    hover_name="geo_id",
    hover_data=["combined_area_effect_width", "combined_area_effect_mean"],
    title="Uncertainty width for residual area effects",
)
fig.update_geos(fitbounds="locations", visible=False)
fig.update_layout(margin={"l": 0, "r": 0, "t": 50, "b": 0})
fig
```

## Interpretation reminder

The residual area effects are on the log-rate scale. If their credible intervals cross zero, that means the model does not strongly identify an additional residual spatial effect for that municipality after accounting for exposures and age controls.
